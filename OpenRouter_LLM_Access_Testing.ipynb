```python
import os
import getpass

if not os.environ.get("OPENROUTER_API_KEY"):
    os.environ["OPENROUTER_API_KEY"] = getpass.getpass(
        "Enter your OpenRouter API key: "
    )

print("✅ OpenRouter API key loaded")
```

```python

import requests
from openai import OpenAI

MODEL_NAME = "openai/gpt-4o-mini"

client = OpenAI(
    base_url="https://openrouter.ai/api/v1", #Send requests to OpenRouter instead of directly to OpenAI
    api_key=os.environ["OPENROUTER_API_KEY"],

)

print(f"✅ Client initialized")
```

```python
basic_response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {
            "role": "user",
            "content": "Who is the POTUS",
        },
    ],
    temperature=0.2,
    top_p=0.9,
    max_tokens=150
)

basic_answer = basic_response.choices[0].message.content
print(basic_answer)
```
